```python
data4={'Name': ['Ronald', 'tom ', 'barney'], 'age': [24, 61, 76]}
data4_df = pd.DataFrame(data4)
print(data4_df.iloc[0])
print(data4_df.iloc[1])
#row indicate position of the row in the DataFrame. The first row has an index of 0, the second row has an index of 1, and so on. You can use the iloc method to access rows by their index position.

print(data4_df.iloc[0,0])  # Accessing the first row, first column
print(data4_df.iloc[1,1])  # Accessing the second row, second column
print(data4_df.iloc[2,0])  # Accessing the third row, first column
print(data4_df.iloc[0,1])  # Accessing the first row, second column
print(data4_df.iloc[1:2]) 
print(data4_df.iloc[0:2,1])
print(data4_df.iloc[0:2,0:2])
print(data4_df.iloc[:,:-1])  # Accessing all rows and all columns except the last one
print(data4_df.iloc[:,1:])  # Accessing all rows and all columns starting from the second one
print(data4_df.iloc[:,-1])
print(data4_df.iloc[0:2,0:2])  # Accessing the first two rows and the first two columns


```
